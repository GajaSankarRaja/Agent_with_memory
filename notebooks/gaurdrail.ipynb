```python
from langchain.agents.middleware import PIIMiddleware
from langchain.messages import HumanMessage
from langchain.agents import create_agent
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os

load_dotenv()


API_KEY = os.getenv("API_KEY")
```

```python
# Middleware:
pil_middleware = PIIMiddleware(
    "email",
    strategy="redact",
    apply_to_input=True,
    apply_to_output=True,
)
```

```python
# Groq Model:
model = ChatGroq(
    model='openai/gpt-oss-120b',
    api_key=API_KEY,
    temperature=0.3
)
```

```python
agent = create_agent(
    model=model,
    #tools=[get_customer_contact],
    middleware=[
        pil_middleware
    ],
)
```

```python
result = agent.invoke({
    "messages": [HumanMessage(content="Return the given email user123@gmail.com")]
})
```
